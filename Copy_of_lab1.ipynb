```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Load Boston Housing dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.boston_housing.load_data()

print(X_train.shape)
print("Input features:", X_train[0])
print("Output target:", y_train[0])

# Select one feature
x_train = X_train[:, 10]
x_test = X_test[:, 10]

# Visualize data
plt.xlabel("Feature 1")
plt.ylabel("Median Price [$K]")
plt.grid(True)
plt.scatter(x_train, y_train, color="green", alpha=0.5)
plt.show()

# Create Sequential Model
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(1, input_shape=(1,)))

model.summary()

# Compile model
model.compile(
    optimizer=tf.keras.optimizers.RMSprop(learning_rate=0.005),
    loss="mse"
)

# Train model
hist = model.fit(
    x_train,
    y_train,
    batch_size=32,
    epochs=100,
    validation_split=0.3
)

# Plot training and validation loss
history = hist.history

plt.plot(history['loss'], label='Train Loss')
plt.plot(history['val_loss'], label='Validation Loss')
plt.legend()
plt.show()

# Predict
x = np.array([3])
y_pred = model.predict(x)
print("Prediction:", y_pred)

# Predict test data
y_pred = model.predict(x_test)

plt.scatter(x_test, y_test, label='Ground Truth', color='green')
plt.scatter(x_test, y_pred, label='Predictions', color='red')
plt.legend()
plt.show()
```

# Multiple linear regression


```python
import tensorflow as tf
import matplotlib.pyplot as plt

# Load dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.boston_housing.load_data()

print(X_train.shape)
print(X_test.shape)

# Create model
new_model = tf.keras.models.Sequential()

new_model.add(
    tf.keras.layers.Dense(120, activation='relu', input_shape=(13,))
)

new_model.add(tf.keras.layers.Dense(1))

new_model.summary()

# Compile
new_model.compile(
    optimizer='rmsprop',
    loss='mse'
)

# Train
new_hist = new_model.fit(
    X_train,
    y_train,
    batch_size=32,
    validation_split=0.3,
    epochs=100
)

# Plot loss
history = new_hist.history

plt.plot(history['loss'], label='Train Loss')
plt.plot(history['val_loss'], label='Validation Loss')
plt.legend()
plt.show()

# Evaluate
test_loss = new_model.evaluate(X_test, y_test)
print("Test Loss:", test_loss)

# Predict
predictions = new_model.predict(X_test[:5])
print(predictions)
```
